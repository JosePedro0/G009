## 1. Formulação do problema

A BVE precisa avaliar como ampliar em 20% sua capacidade de bunkering, conciliando janelas dos navios, frota, estoques, deslocamentos e carregamento. O estudo apoiará a escolha entre mudanças de programação, disponibilidade de barcaças e capacidade do terminal. A simulação permitirá avaliar filas, variabilidade e utilização antes de modificar a operação.

O caso de Costa e Mesquita (2023) descreve seis barcaças de 1.500 a 3.000 t, dois berços de carregamento e programação duas vezes ao dia. Esses valores são referências do artigo, não parâmetros confirmados da BVE. Frota, berços e regime efetivos serão confirmados no diagnóstico.



## 2. Objetivo e escopo

**Objetivo:** avaliar alternativas capazes de elevar em pelo menos 20% a capacidade sustentável de entrega, preservando o nível de serviço acordado com a BVE.

Capacidade será o maior volume médio diário entregue sob um perfil de demanda fixo, com pelo menos 95% dos pedidos elegíveis concluídos na janela física e sem crescimento persistente da carteira pendente. O limiar de 95% é uma premissa proposta. A meta será C1/C0 ≥ 1,20; volume histórico realizado não será automaticamente tratado como capacidade.

| Pergunta do estudo | Indicador e unidade |
|:--|:--|
| Qual capacidade é sustentável e qual alternativa alcança a meta? | Volume entregue (t/dia); ganho relativo (%) |
| O serviço é preservado? | Pedidos pontuais e não atendidos (%); atraso comercial (h) |
| Onde estão os gargalos? | Espera média e P95 (h); utilização por recurso (%) |
| A operação permanece estável? | Carteira pendente (pedidos e t) ao longo do tempo |

Incluem-se demanda, abastecimento, carregamento e PCO. Chegada, atracação e saída dos navios serão entradas externas; não se simulará toda a operação de carga portuária. O cenário-base reproduzirá recursos, políticas e calendário confirmados da BVE. Para validação será reproduzido um período histórico; para capacidade, propõem-se 90 dias de observação após aquecimento, ampliáveis se necessários à estabilidade e sazonalidade.

Serão comparadas programação atual versus menor folga de janela, maior disponibilidade da frota e maior capacidade de carregamento, isoladamente e em combinações selecionadas. Essas alternativas são propostas da equipe e serão confrontadas com as sugestões da carta-convite; seus níveis serão definidos no Plano de Experimentos.

Excluem-se integração em produção, engenharia de instalações e otimização exata de grande porte, para respeitar as 240 horas disponíveis. Reposição de combustível será uma entrada externa, mantendo o controle de estoque.



## 3. Método

O diagnóstico produzirá o modelo conceitual e o pedido de dados. O tratamento gerará uma base documentada; a implementação produzirá uma simulação em Python/SimPy e análises em Jupyter. Verificação de regras e balanços precederá a validação com um período histórico separado e revisão operacional. Experimentos compararão cenários com sementes correspondentes e intervalos de confiança de 95%; sensibilidade avaliará hipóteses frágeis. Costa e Mesquita (2023), seção 3, fundamentam a descrição operacional; o presente estudo propõe simulação, enquanto o artigo desenvolve otimização do sequenciamento.
